```python
import os
import random
import shutil

random.seed(42)

root = r"D:\Python\SemCom\saved_data\deepglobe_64"

folders = ["train", "valid", "test"]

train_target = 26800
test_target = 200

unused_dir = os.path.join(root, "unused")
os.makedirs(unused_dir, exist_ok=True)

# 支援的副檔名
exts = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")

# --------------------------------------------------
# 收集所有圖片
# --------------------------------------------------
all_imgs = []

for folder in folders:
    folder_path = os.path.join(root, folder)

    for f in os.listdir(folder_path):
        if f.lower().endswith(exts):
            all_imgs.append(os.path.join(folder_path, f))

print(f"Total images : {len(all_imgs)}")

# --------------------------------------------------
# 隨機打亂
# --------------------------------------------------
random.shuffle(all_imgs)

train_imgs = all_imgs[:train_target]
test_imgs = all_imgs[train_target:train_target+test_target]
unused_imgs = all_imgs[train_target+test_target:]

print("Train :", len(train_imgs))
print("Test  :", len(test_imgs))
print("Unused:", len(unused_imgs))

# --------------------------------------------------
# 清空 train valid test
# --------------------------------------------------
for folder in folders:
    folder_path = os.path.join(root, folder)

    for f in os.listdir(folder_path):
        file_path = os.path.join(folder_path, f)

        if os.path.isfile(file_path):
            os.remove(file_path)

# --------------------------------------------------
# 搬回 train
# --------------------------------------------------
for src in train_imgs:
    shutil.move(
        src,
        os.path.join(root, "train", os.path.basename(src))
    )

# --------------------------------------------------
# 搬回 test
# --------------------------------------------------
for src in test_imgs:
    shutil.move(
        src,
        os.path.join(root, "test", os.path.basename(src))
    )

# --------------------------------------------------
# 剩下全部搬到 unused
# --------------------------------------------------
for src in unused_imgs:
    shutil.move(
        src,
        os.path.join(unused_dir, os.path.basename(src))
    )

print("Finished!")
```
