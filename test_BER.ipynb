```python
import numpy as np
from pathlib import Path
from scipy.io import loadmat
import torch
```

```python
i=0
tx_data = loadmat(f'D:/matlab/result_20260413/TX_result/result_B_{i:03d}.mat')['B'].flatten()
rx_data = loadmat(f'D:/matlab/result_20260413/RX_result/result_B_bits_{i:03d}.mat')['bits'].flatten()

tx = tx_data
rx = rx_data

win_size = 100   # window 大小（自己調）
tx_win = tx[:win_size]

best_score = -1
best_idx = 0

# === sliding window ===
for i in range(len(rx) - win_size + 1):
    window = rx[i:i+win_size]

    score = 0
    for j in range(win_size):
        if window[j] == tx_win[j]:
            score += 1   # 一個 bit 對就 +1

    if score > best_score:
        best_score = score
        best_idx = i

print("best index:", best_idx)
print("score:", best_score)

L = len(tx)
rx_extract = rx[best_idx:best_idx+L]

tx_bits = tx.astype(int)
rx_bits = rx_extract.astype(int)

bit_errors = np.sum(tx_bits != rx_bits)
ber = bit_errors / L

print("BER:", ber)
```
