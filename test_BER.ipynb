```python
bits = 8
num = 0
ber_sum = 0
error_num = 0
print("bits=",bits)
for i in range(0,199):
    # 讀檔
    tx_data = loadmat(f'D:/matlab/result_20260317/latent_bits_M/bits_{bits}/sample_{i:03d}.mat')
    
    rx_data = loadmat(f'D:/matlab/result_20260317/latent_bits_M/bits_{bits}/result/result_{i:03d}.mat')
    
    # rx_data = loadmat(f'D:/matlab/result_20260317/latent_bits_M/bits_{bits}/result/result_{i-1:03d}.mat')
    # 取出變數
    tx_bits = tx_data['bits'].flatten().astype(np.uint8)
    rx_bits = rx_data['bits'].flatten().astype(np.uint8)

    # 長度檢查
    # print("tx length:", len(tx_bits))
    # print("rx length:", len(rx_bits))

    # 如果長度不同，先截成相同長度
    L = min(len(tx_bits), len(rx_bits))
    tx_bits = tx_bits[:L]
    rx_bits = rx_bits[:L]

    # 計算 BER
    bit_errors = np.sum(tx_bits != rx_bits)
    ber = bit_errors / L
    ber_sum = ber_sum + ber
    # print("Bit errors:", bit_errors)
    # print("Total bits:", L)
    # print("BER:", ber)
    if ber != 0:
        error_num += 1
        print("i=",i," ber=",ber);
ave_ber = ber_sum/200
print(ave_ber)
print(error_num)
```
